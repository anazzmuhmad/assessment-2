Q13

```python
student["Average_Technical_Score"] = student[
    [
        "Python_Score",
        "SQL_Score",
        "Pandas_Score",
        "NumPy_Score"
    ]
].mean(axis=1)
```

```python
student["Performance"] = np.select(
    [ student["Average_Technical_Score"] >= 90,
        student["Average_Technical_Score"] >= 75,
        student["Average_Technical_Score"] >= 60 ],
    [ "Excellent",
        "Good",
        "Average" ],
    default = "Needs Improvement"
)
```

```python
student.head(80)
```

Q14

```python
department_result = student.groupby("Department").agg(
    Number_of_Students=("Student_ID", "count"),
    Average_Final_Score=("Final_Score", "mean"),
    Maximum_Final_Score=("Final_Score", "max"),
    Minimum_Final_Score=("Final_Score", "min")
)

department_result = department_result.sort_values(
    "Average_Final_Score",
    ascending=False
)
department_result.round(2)
```

Section C

Q16

A

```python
student.isnull().sum()
```

```python
columns = [
    "Age",
    "Attendance",
    "Python_Score",
    "SQL_Score",
    "Pandas_Score",
    "NumPy_Score",
    "Final_Score"
]

for column in columns:
    student[column] = student[column].fillna(
        student[column].mean()
    )
```

B

```python
student["Average_Score"] = student[
    [
        "Python_Score",
        "SQL_Score",
        "Pandas_Score",
        "NumPy_Score"
    ]
].mean(axis=1)
```

```python
student["Pass_Status"] = np.where(
    student["Average_Score"] >= 50,
    "Pass",
    "Fail"
)
```

C

```python
student.sort_values(
    "Average_Score",
    ascending=False
).head(5)[
    ["Student_ID", "Name", "Average_Score"]
]
```

```python
student.groupby(
    "Department"
)["Average_Score"].mean().sort_values(
    ascending=False
)
```

```python
student.groupby(
    "Department"
)["Attendance"].mean().round(2)
```

```python
student["Pass_Status"].value_counts()
```

D

```python
summary = student.groupby(
    "Department"
).agg(
    Total_Students=("Student_ID", "count"),
    Average_Score=("Average_Score", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
    Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
).reset_index()

summary = summary.round(2)

summary
```
